# Change Data Feed (CDF)

```sql
%sql

CREATE OR REPLACE TABLE main.silver.account_transactions
(
    account_id INT,
    customer_name STRING,
    balance DECIMAL(10,2)
)
TBLPROPERTIES (
    delta.enableChangeDataFeed = true
);
```

```sql
%sql

SHOW TBLPROPERTIES main.silver.account_transactions;
```

```sql
%sql

INSERT INTO main.silver.account_transactions VALUES
(101,'Rajesh Kumar',45000),
(102,'Priya Sharma',68000),
(103,'Arjun Reddy',52000);
```

```sql
%sql

DESCRIBE HISTORY main.silver.account_transactions;
```

# Make some changes

```sql
%sql

UPDATE main.silver.account_transactions
SET balance = 55000
WHERE account_id = 101;
```

```sql
%sql

INSERT INTO main.silver.account_transactions
VALUES
(104,'Ananya Gupta',78000);
```

```sql
%sql

DELETE
FROM main.silver.account_transactions
WHERE account_id = 102;
```

# Read the Change Data Feed ⭐

```sql
%sql

SELECT *
FROM table_changes(
    'main.silver.account_transactions',
    1
);
```
