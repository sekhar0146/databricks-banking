```sql
%sql

CREATE OR REPLACE TABLE banking.bronze.customer_master
(
    customer_id     INT,
    customer_name   STRING,
    city            STRING,
    balance         DECIMAL(12,2),
    customer_type   STRING
)
USING DELTA;
```

```sql
%sql

CREATE OR REPLACE TABLE banking.bronze.account_master
(
    account_id      INT,
    customer_id     INT,
    account_type    STRING,
    branch          STRING,
    balance         DECIMAL(12,2)
)
USING DELTA;
```

```sql
%sql

CREATE OR REPLACE TABLE banking.bronze.transaction_master
(
    transaction_id      INT,
    account_id          INT,
    transaction_date    DATE,
    transaction_type    STRING,
    amount              DECIMAL(12,2)
)
USING DELTA;
```
