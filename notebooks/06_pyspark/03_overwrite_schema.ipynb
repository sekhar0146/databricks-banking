```python
from pyspark.sql.types import *

customer_schema_v2 = StructType([
    StructField("customer_id", IntegerType(), False),
    StructField("customer_name", StringType(), False),
    StructField("city", StringType(), True),
    StructField("balance", DecimalType(12,2), True),
    StructField("email", StringType(), True)
])

df = (
    spark.read
         .option("header", "true")
         .schema(customer_schema_v2)
         .csv("/Volumes/banking/bronze/landing_volume/customers_v2.csv")
)
```

```python
df = df.withColumnRenamed("customer_name", "cust_name")
```

```python
display(df)
```

```python
df.write \
  .mode("overwrite") \
  .option("mergeSchema","true") \
  .saveAsTable("banking.bronze.customer_master")
```

```python
%sql
select * from banking.bronze.customer_master;
```

## overwriteSchema

```python
from pyspark.sql.types import *

customer_schema_v2 = StructType([
    StructField("customer_id", IntegerType(), False),
    StructField("customer_name", StringType(), False),
    StructField("city", StringType(), True),
    StructField("balance", DecimalType(12,2), True),
    StructField("email", StringType(), True)
])

df2 = (
    spark.read
         .option("header", "true")
         .schema(customer_schema_v2)
         .csv("/Volumes/banking/bronze/landing_volume/customers_v2.csv")
)
```

```python
display(df2)
```

```python
df2 = df2.withColumnRenamed("customer_name", "cust_name")
```

```python
df2.write \
  .mode("overwrite") \
  .option("overwriteSchema", "true") \
  .saveAsTable("banking.bronze.customer_master")
```

```python
display(df2)
```

```python
%sql
select * from banking.bronze.customer_master;
```
