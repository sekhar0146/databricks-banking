```python
df = spark.read.table("banking.bronze.customer_master")
```

```python
display(df)
```

```python
# Select specific columns and filter rows where 'balance' > 50000
df_transformed = df.select("customer_id", "cust_name", "balance").filter(df.balance > 50000)

# Group by 'city' and count customers
df_grouped = df.groupBy("city").count()

# Display the transformed DataFrame
display(df_transformed)

# Display the grouped DataFrame
display(df_grouped)
```

```python
df.filter("balance >= 70000").show()
```

```python
from pyspark.sql.functions import col
df.select(
    col("cust_name").alias("Customer"),
    col("balance").alias("Current_Balance")
).show()
```

```python
df.orderBy(col("balance").desc()).show()
```

### Combine prev code blocks

```python
df.select(
    "customer_id",
    "cust_name",
    "balance"
).filter(
    col("balance") >= 50000
).orderBy(
    col("balance").desc()
).show()
```
