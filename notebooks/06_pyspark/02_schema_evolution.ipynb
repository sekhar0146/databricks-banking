```python
from pyspark.sql.types import *

customer_schema_v2 = StructType([
    StructField("customer_id", IntegerType(), False),
    StructField("customer_name", StringType(), False),
    StructField("city", StringType(), True),
    StructField("balance", DecimalType(12,2), True),
    StructField("email", StringType(), True)
])

df = (
    spark.read
         .format("csv")
         .option("header", "true")
         .schema(customer_schema_v2)
         .load("/Volumes/banking/bronze/landing_volume/customers_v2.csv")
)
```

```python
display(df)
```

## try writing it without schema evolution.

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("banking.bronze.customer_master")
```

## fix it by enabling schema evolution

```python
df.write \
    .format("delta") \
    .mode("overwrite") \
    .option("mergeSchema", "true") \
    .saveAsTable("banking.bronze.customer_master")
```
