```python
# Bronze Layer — Raw Data Ingestion
# Source: samples.bakehouse (Databricks built-in)

# Step 1: Read source data
df_transactions = spark.table("samples.bakehouse.sales_transactions")

# Preview
display(df_transactions)
```

```python
# Step 2: Understand the data
print(f"There are {df_transactions.count()} records in the DataFrame.")
print(f"Columns: {', '.join(df_transactions.columns)}")
print(f"Schema:")
df_transactions.printSchema()
```

```python
# Step 3: Write to Bronze Delta Table
(df_transactions
    .write
    .format("delta")
    .mode("overwrite")
    .partitionBy("dateTime")
    .saveAsTable("workspace.default.bronze_transactions")
)

print("Bronze table created successfully!")
```
